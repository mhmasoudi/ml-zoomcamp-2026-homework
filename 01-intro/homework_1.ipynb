```python
X = df[df['origin'] == 'Asia'][["vehicle_weight", "model_year"]].head(7).values
```

```python
X
```

```python
X_tra = X.T
```

```python
X_tra
```

```python
X_tra.dot(X)
```

```python
XTX = X_tra.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX_inv
```

```python
XTX_inv @ XTX
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```
